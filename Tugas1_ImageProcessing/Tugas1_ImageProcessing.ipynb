```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Baca citra grayscale (pastikan file ada di direktori yang sama)
img = cv2.imread('Gambar.jpg', cv2.IMREAD_GRAYSCALE)

# Tampilkan informasi citra
print('Ukuran citra:', img.shape)
print('Tipe data:', img.dtype)
```

```python
a = 100
```

```python
# Konversi ke int16 agar tidak overflow saat ditambah
img_float = img.astype(np.int16)
brightened = img_float + a

# Clipping ke rentang 0-255
brightened = np.clip(brightened, 0, 255)

# Kembalikan ke uint8
brightened = brightened.astype(np.uint8)
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title('Citra Asli')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(brightened, cmap='gray')
plt.title(f'Citra setelah pencerahan a = {a}')
plt.axis('off')

plt.show()
```
